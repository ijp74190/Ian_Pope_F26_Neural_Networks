Q1. NLP Preprocessing Pipeline

```python
#Import required libraries
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
```

```python
#Download required resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
#Create the preprocessing function
def nlp_preprocessing(sentence):

  #Tokenize the sentence
  tokens = word_tokenize(sentence)
  print('Original Tokens:', tokens)

  #Remove stop words (and punctuation)
  stop_words = set(stopwords.words('english'))
  filtered_tokens = [word for word in tokens
                     if word.lower() not in stop_words
                     and word not in string.punctuation]
  print('Tokens Without Stopwords:', filtered_tokens)

  #Apply stemming
  stemmer = PorterStemmer()
  stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
  print('Stemmed Tokens:', stemmed_tokens)
```

```python
#Call preprocessing on the sentence
sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."

nlp_preprocessing(sentence)
```

Q2. Scaled Dot-Product Attention

```python
#Import required libraries
import numpy as np
```

```python
#Create a function that performs the scaled dot-product attention
def scaled_dot_product_attention(Q, K, V):

  #Calculate the dot product of Q and K
  dot_product = np.dot(Q, K.T)

  #Divide scores by square root of d
  d = K.shape[1]
  scaled_dot_product = dot_product / np.sqrt(d)

  #Apply softmax to the scaled dot product
  exp_scores = np.exp(scaled_dot_product)
  attention_weights = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)

  #Calculate the output
  output = np.dot(attention_weights, V)

  #Print results
  print("Attention Weights:")
  print(attention_weights)
  print()

  print("Final Output:")
  print(output)
```

```python
#Define the inputs
Q = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
K = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
V = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])

#Call the function
scaled_dot_product_attention(Q, K, V)
```

1. Why do we divide the attention score by √d in the scaled dot-product attention formula?

Dividing by the sqrt(d) scales down the attention scores prevents the dot product from getting too large for the softmax function to use appropriately, as the dot product can scale quickly as d increases.  

2. How does self-attention help the model understand relationships between words in a sentence?

Self-attention helps models understand relationships between words by allowing the model to learn what parts of a sentence provide the context for the word in question. This removes ambiguity for nouns such as 'it' and words with multiple meanings such as 'bank'.
